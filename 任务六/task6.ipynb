```python
# 只安装兼容的版本，避免版本冲突
!pip install transformers==4.37.2 torch==2.1.0

from transformers import AutoTokenizer, AutoModelForCausalLM, pipeline

# 使用和你作业要求一致的模型
model_name = "uer/gpt2-chinese-cluecorpussmall"

# 自动加载模型和分词器，自动适配
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name)

# 配置续写，强制使用CPU运行，彻底避开所有分布式模块
generator = pipeline(
    "text-generation",
    model=model,
    tokenizer=tokenizer,
    max_length=150,
    temperature=0.7,
    repetition_penalty=1.1,
    do_sample=True,
    pad_token_id=tokenizer.eos_token_id,
    device=-1  # 强制CPU运行，不会触发任何分布式/FSDP相关报错
)

# 学号末位0 固定开头句子
prompt = "如果我拥有一台时间机器"

# 执行续写并输出结果
result = generator(prompt)
print("文本续写结果：\n", result[0]["generated_text"])
```
