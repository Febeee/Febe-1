# 结巴分词实践

## 结巴分词的三种模式

```python
import jieba
# 精确模式
text = "我来到北京清华大学"
result = jieba.cut(text, cut_all=False)
print("精确模式: ", "/".join(result))
```

### 全模式

```python
import jieba
# 全模式
text = "我来到北京清华大学"
result = jieba.cut(text, cut_all=True)
print("全模式: ", "/".join(result))
```

### 搜索引擎模式

```python
import jieba
# 搜索引擎模式
text = "小明硕士毕业于中国科学院计算所，后在日本京都大学深造"
result = jieba.cut_for_search(text)
print("搜索引擎模式: ", "/".join(result))
```

## 载入自定义词典

```python
import jieba
# 加载自定义词典
jieba.load_userdict("userdict.txt")
# 测试分词
text = "李亦非即将是自然语言处理方面的高手"
result = jieba.cut(text, cut_all=False, HMM=False)
print("自定义词典分词: ", "/".join(result))
```
